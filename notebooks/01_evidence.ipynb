# Reference equation checks and actual demand totals

```python
import sys
from pathlib import Path

import pandas as pd

sys.path.insert(0, str(Path("src").resolve()))
from water_demand.physics import penman_monteith, saturation_pressure

ea = (saturation_pressure(12.3) * 0.84 + saturation_pressure(21.5) * 0.63) / 2
case = penman_monteith(21.5, 12.3, 73.5, 2.078, 22.07, 100.1, 50.8, 187, 100, ea)
assert abs(float(case["et0_pm_mm"]) - 3.88) < 0.03
float(case["et0_pm_mm"])
```

```python
daily = pd.read_csv("reports/daily-water-demand.csv")
monthly = pd.read_csv("reports/monthly-water-demand.csv")
assert len(daily) == 1098
assert abs(daily.et0_pm_mm.sum() - monthly.et0_pm_mm.sum()) < 1e-8
assert (abs(daily["etc_scenario_kc_0.8_mm"] - 0.8 * daily.et0_pm_mm) < 1e-10).all()
pd.read_csv("reports/method-comparison.csv")
```
